```python

import math
v = float(input('velocidad inicial m/s '))
angulo = float(input('angulo de lanzamiento (°)'))
rad = math.radians(angulo)
vx = lambda v, rad: v * math.cos(rad)
vy = lambda v, rad: v * math.sin(rad)
tiempo = lambda vy: (2 * vy) / 9.81
velx = vx(v, rad)
vely = vy(v, rad)
t = tiempo(vely)
print('resultados')
print('velocidad horizontal', round(velx, 2), 'm/s')
print('velocidad vertical', round(vely, 2), 'm/s')
print('tiempo de vuelo', round(t, 2), 's')
```

```python
import math
angulos = [15, 30, 45, 60, 75]
v = 30
g = 9.81
alcance = lambda ang: (v**2 / g) * math.sin(2 * math.radians(ang))
distancias = list(map(alcance, angulos))
mayores70 = list(filter(lambda d: d > 70, distancias))
print('angulos', angulos)
print('distancias calculadas', [round(d,1) for d in distancias])
print('lanzamientos destacados (>70 m)', [round(d,1) for d in mayores70])
```

```python
from functools import reduce
distancias = [35.7, 65.3, 91.8, 79.5, 48.2]
suma = reduce(lambda a, b: a + b, distancias)
promedio = suma / len(distancias)
maxima = max(distancias)
mayores_prom = [d for d in distancias if d > promedio]

print('distancias', distancias)
print('distancia promedio', round(promedio,1), 'm')
print('distancia máxima', maxima, 'm')
print('lanzamientos sobre el promedio', mayores_prom)
```
