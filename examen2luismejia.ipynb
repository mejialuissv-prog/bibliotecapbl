```python

ingresos = []

n = int(input('Cuántos productos ingresaras '))
for i in range(n):
    print('Producto', i+1)
precio = float(input(' precio '))
cantidad = int(input(' cantidad de venta '))

ingreso = precio * cantidad
ingresos.append(ingreso)

print('Ingresos por producto', ingresos)

total = sum(ingresos)
print('Ingreso total', total)
```

```python
ingresos = [12000, 4500, 16000, 1500, 125]
destacados = []

for ingreso in ingresos:
    if ingreso > 10000:
        destacados.append(ingreso)
print('ingresos', ingresos)
print('Productos destacados' destacados)
```
