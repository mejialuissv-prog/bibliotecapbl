hospitales = []
nombres = ['hospital A', 'hospital B', 'hospital C', 'hospital D']

for i in range(len(nombres)):
    print('ingrese datos para ' + nombres[i])
    meses = []
    for j in range(6):
        cantidad = int(input('  ' + nombres[i] + ' - mes ' + str(j+1) + ': '))
        meses.append(cantidad)
    hospitales.append(meses)

print('datos obtenidos:')
print(hospitales)
