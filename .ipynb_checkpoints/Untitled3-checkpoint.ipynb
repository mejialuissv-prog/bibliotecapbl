```python
import math
def operacion(a,b):
    r2 =(a**2)+(b**2)
    r = math.sqrt(r2)
    return r 
```

```python
A = int(input('ingrese lado A:'))
B = int(input('ingrese lado B:'))
r = operacion (A,B)
print(f' la hipotenusa de los lados a={A} y b={B} es: {round(r,2)}')
```
