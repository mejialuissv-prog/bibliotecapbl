```python
puntos = []

equipos = 8
jornadas = 5


for i in range(equipos):
    print('equipo', i + 1)
    equipo = []
    
    for j in range(jornadas):
        dato = int(input('Ingrese los puntos de la jornada ' + str(j + 1)))
        equipo.append(dato)
    
    puntos.append(equipo)

print('\nlista completa de puntos')
print(puntos)


totales = []

for i in range(equipos):
    suma = 0
    for j in range(jornadas):
        suma = suma + puntos[i][j]
    totales.append(suma)

print('\ntotal de puntos por equipo')
for i in range(equipos):
    print('Equipo', i + 1, totales[i])


print('\npromedio de puntos por partido')
promedios = []

for i in range(equipos):
    promedio = totales[i] / jornadas
    promedios.append(promedio)
    print('equipo', i +1, promedio)

mayor = totales[0]

for i in range(equipos):
    if totales[i] > mayor:
        mayor = totales[i]

print('\nequipo líder')
for i in range(equipos):
    if totales[i] == mayor:
        print('equipo', i + 1, 'con', mayor, 'puntos')
```
